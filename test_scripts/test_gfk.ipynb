```python
import requests
from airflow.models import Variable  # можно убрать, если хочешь руками токен вставить

# ===== ВАРИАНТ 1: Тянуть токен из Airflow Variable =====
try:
    api_key = Variable.get("gfk_token")
except Exception as e:
    print("⚠️ Не удалось получить токен из Airflow Variable:", e)
    api_key = None

# ===== ВАРИАНТ 2: Вставить токен руками =====
# api_key = "ТОКЕН_ОТ_GFK"

if not api_key:
    raise ValueError("❌ Токен не найден, задай его в Airflow Variable gfk_token или вставь в код!")

url = "https://startrack.mi.gfk.com/api/v1/Reports?publishedSinceDaysOffset=30&page=1&pageSize=5"

headers = {
    "Authorization": f"Bearer {api_key}",
    "Accept": "application/json"
}

print("🔎 Делаем запрос к API:", url)

response = requests.get(url, headers=headers)

print("➡️ Status code:", response.status_code)

if response.status_code == 200:
    print("✅ Успешный ответ!")
    print(response.json())  # напечатает первые данные
else:
    print("❌ Ошибка при запросе")
    print(response.text)
```
